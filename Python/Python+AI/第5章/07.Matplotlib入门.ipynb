```python
import random
import matplotlib.pyplot as plt

# 设置 Jupyter 内联输出为 Retina 高清格式
%config InlineBackend.figure_format = 'retina'

# 设置中文字体 - MacOS使用该字体
plt.rcParams["font.sans-serif"] = ["Arial Unicode MS"]
plt.rcParams["axes.unicode_minus"] = False

# 绘制折线图
x = [i for i in range(1, 25)]
y_bj = [random.randint(5, 15) for i in x]
y_hz = [random.randint(10, 23) for i in x]

# 设置画布大小，宽，高
plt.figure(figsize=(10, 6))

# 折线图
plt.plot(x, y_bj, label="北京", marker='o', markersize=4)
plt.plot(x, y_hz, label="杭州", marker='s', markersize=4)

# 标题
plt.title("气温变化折线图", fontsize=16)
# X、Y轴标签
plt.xlabel("时间")
plt.ylabel("气温")

# X、Y轴刻度
plt.xticks(x)
y_ticks = [i for i in range(0, 26)]
plt.yticks(y_ticks)

# 显示网格
plt.grid(linestyle="--", alpha=0.3)

# 显示图例
plt.legend(loc="upper right")

# 显示图表
plt.show()
```
