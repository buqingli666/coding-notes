```python
import pandas as pd

# 创建 DataFrame
df1 = pd.DataFrame([
    {'姓名': '王林', '语文': 80, '英语': 90, '数学': 88},
    {'姓名': '慕婉', '语文': 92, '英语': 81, '数学': 93},
    {'姓名': '十三', '语文': 87, '英语': 83, '数学': 78}
])

df2 = pd.DataFrame({
    '姓名': ['王林', '慕婉', '十三'],
    '语文': [80, 92, 87],
    '英语': [90, 81, 83],
    '数学': [88, 93, 78]
})

df3 = pd.DataFrame([
    ['王林', 80, 90, 88],
    ['慕婉', 92, 81, 93],
    ['十三', 87, 83, 78]
], columns=['Name', 'Chinese', 'English', 'Math'])

df4 = pd.DataFrame([
    ('王林', 80, 90, 88),
    ('慕婉', 92, 81, 93),
    ('十三', 87, 83, 78)
], columns=['姓名', '语文', '英语', '数学'], index=['a', 'b', 'c'])

# DataFrame 常见属性
# index：索引
print("index: 索引 ->", df1.index.tolist())

# columns：列名
print("columns: 列名 ->", df4.columns.tolist())

# values：值
print("values: 值 ->", df4.values.tolist())

# size：元素个数
print("size: 元素个数 ->", df4.size)

# dtypes：数据类型
print("dtypes: 数据类型 ->", df4.dtypes)

# shape：维度(行,列)
print("shape: 维度(行,列) ->", df4.shape)
```
